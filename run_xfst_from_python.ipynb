### Run xfst from python

make sure to have your xfst application and your xfst file in the same directory

```python
import subprocess

directory = r"C:\Users\ueffi\OneDrive\Desktop\Kayardild_AP_KevinUeffing2026"
filename = "kayardild.xfst"
application = r"C:\Users\ueffi\OneDrive\Desktop\ComputationalMorphology\XFST\bin\xfst.exe"
```

```python
commands = [f"source < {directory}/{filename};"]
```

```python
import os
print(os.path.exists(application))
print(os.path.isfile(application))
```

```python
result = subprocess.run([application],
                    input="\n".join(commands),
                    text = True,
                    capture_output=True,
                    cwd=directory)
print(result.stdout)
print(result.stderr)
```

```python
with open(f"{directory}/kayardild_output.txt", "r") as f:
    output = f.readlines()

output
```

```python
with open(f"{directory}/kayardild_goals.txt", "r") as f:
    goal = f.readlines()
```

```python
import difflib
```

```python
diff = list(difflib.ndiff(output, goal))
changed = [d for d in diff if d.startswith('+') or d.startswith('-')]
if not changed:
    print(f"All {len(goal)} lines match. No differences found.")
else:
    print(f"{len(changed)} differing line(s) found:")
    for d in changed:
        print(d)
```
